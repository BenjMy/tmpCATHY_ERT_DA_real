# Uncompacted

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d.art3d import Poly3DCollection
figdir = '../figures/'

# Load electrodes
elecs_raw = np.genfromtxt('../data/elecs72bh-def.csv', delimiter=",", skip_header=1)
el_x = elecs_raw[:, 1]
el_y = elecs_raw[:, 2]
el_z = elecs_raw[:, 3]

# Calcolo del centro medio degli elettrodi (media delle coordinate x, y e z)
center_x = np.mean(el_x)
center_y = np.mean(el_y)
center_z = np.mean(el_z)

# Dimensioni del cubo 3x3
half_cube_size = 1.2  # metà della dimensione del cubo (3x3)
z = 2

# Variabili per le dimensioni del parallelepipedo (modifica qui per cambiarne la forma)
x_min = 0.2
x_max = 0.7
y_min = -0.75
y_max = 1.75
z_min = -.22
z_max = 0

# Definizione dei vertici del cubo centrato sulla base degli elettrodi
verts_cube = [
    [[center_x - half_cube_size, center_y - half_cube_size, 0], [center_x + half_cube_size, center_y - half_cube_size, 0], 
     [center_x + half_cube_size, center_y + half_cube_size, 0], [center_x - half_cube_size, center_y + half_cube_size, 0]],  # Base superiore
    [[center_x - half_cube_size, center_y - half_cube_size, -z], [center_x + half_cube_size, center_y - half_cube_size, -z], 
     [center_x + half_cube_size, center_y + half_cube_size, -z], [center_x - half_cube_size, center_y + half_cube_size, -z]],  # Base inferiore
    [[center_x - half_cube_size, center_y - half_cube_size, 0], [center_x + half_cube_size, center_y - half_cube_size, 0], 
     [center_x + half_cube_size, center_y - half_cube_size, -z], [center_x - half_cube_size, center_y - half_cube_size, -z]],  # Lato X=-1.5
    [[center_x + half_cube_size, center_y - half_cube_size, 0], [center_x + half_cube_size, center_y + half_cube_size, 0], 
     [center_x + half_cube_size, center_y + half_cube_size, -z], [center_x + half_cube_size, center_y - half_cube_size, -z]],  # Lato X=1.5
    [[center_x + half_cube_size, center_y + half_cube_size, 0], [center_x - half_cube_size, center_y + half_cube_size, 0], 
     [center_x - half_cube_size, center_y + half_cube_size, -z], [center_x + half_cube_size, center_y + half_cube_size, -z]],  # Lato Y=1.5
    [[center_x - half_cube_size, center_y + half_cube_size, 0], [center_x - half_cube_size, center_y - half_cube_size, 0], 
     [center_x - half_cube_size, center_y - half_cube_size, -z], [center_x - half_cube_size, center_y + half_cube_size, -z]]   # Lato Y=-1.5
]

# Definizione dei vertici del quadrato 2x2 centrato a (center_x, center_y, z=0)
verts_square = [
    [[center_x - 1, center_y - 1, 0], [center_x + 1, center_y - 1, 0],
     [center_x + 1, center_y + 1, 0], [center_x - 1, center_y + 1, 0]]  # Base
]

# Aggiungere i punti FDR sensor (con z = -15, -30, -50)
fdr_points = np.array([
    [center_x, center_y, -.15],  # Primo punto FDR sensor
    [center_x, center_y, -.30],  # Secondo punto FDR sensor
    [center_x, center_y, -.50]   # Terzo punto FDR sensor
])

# Plot
fig = plt.figure(figsize=(5,6))
ax = fig.add_subplot(111, projection='3d')

# Aggiungi il cubo al grafico
ax.add_collection3d(Poly3DCollection(verts_cube, facecolors='sienna', edgecolors='black', linewidths=.8, alpha=0.2))

# Aggiungi il quadrato 2x2 al grafico
ax.add_collection3d(Poly3DCollection(verts_square, facecolors='b', edgecolors='b', linewidths=1.5, alpha=0.1, label="infiltration area"))

# Aggiungi gli elettrodi
ax.scatter(el_x, el_y, el_z, marker='o', color='r', s=5, label="electrode")

# Aggiungi i punti FDR sensor
ax.scatter(fdr_points[:, 0], fdr_points[:, 1], fdr_points[:, 2], color='k', marker='*', s=20, label="FDR sensor")

# Calcola la coordinata centrale di x nel parallelepipedo
x_center = (x_min + x_max) / 2
ax.plot([x_center, x_center], [y_min, y_max], [0, 0], color='magenta', linewidth=2, label="ERT line")


# Cosmetic settings
ax.set_xlim([-.8, 1.8])
ax.set_ylim([-.8, 1.8])
ax.set_zlim([-2, 0]) 
ax.set_xlabel("x (m)",fontsize=9)
ax.set_ylabel("y (m)",fontsize=9)
ax.set_zlabel("depth (m)",fontsize=9)
#plt.title('NC: no compaction',fontsize=10)
ax.legend(
    loc='upper right',        # Posizione di base della legenda
    fontsize=10,
    frameon=True,
    borderpad=0.2,
    labelspacing=0.3,
    bbox_to_anchor=(1.1, 1.1)   # Spostare la legenda in alto a destra, un po' sopra
)
plt.subplots_adjust(left=0.1, right=0.9, top=0.9, bottom=0.1)
fig.savefig(figdir+ 'scheme_NOcomp.png', dpi=200,)
            #transparent=True)
plt.show()
```

# High Compacted

```python
# Variabili per le dimensioni del parallelepipedo (modifica qui per cambiarne la forma)
x_min = 0.2
x_max = 0.7
y_min = -0.75
y_max = 1.75
z_min = -.22
z_max = 0

# Definizione dei vertici del parallelepipedo rettangolare basato sulle variabili sopra
verts_parallelepiped = [
    [[x_min, y_min, z_max], [x_max, y_min, z_max], [x_max, y_max, z_max], [x_min, y_max, z_max]],  # Base superiore
    [[x_min, y_min, z_min], [x_max, y_min, z_min], [x_max, y_max, z_min], [x_min, y_max, z_min]],  # Base inferiore
    [[x_min, y_min, z_max], [x_max, y_min, z_max], [x_max, y_min, z_min], [x_min, y_min, z_min]],  # Lato X=x_min
    [[x_max, y_min, z_max], [x_max, y_max, z_max], [x_max, y_max, z_min], [x_max, y_min, z_min]],  # Lato X=x_max
    [[x_max, y_max, z_max], [x_min, y_max, z_max], [x_min, y_max, z_min], [x_max, y_max, z_min]],  # Lato Y=y_max
    [[x_min, y_max, z_max], [x_min, y_min, z_max], [x_min, y_min, z_min], [x_min, y_max, z_min]]   # Lato Y=y_min
]


# Plot
fig = plt.figure(figsize=(5,6))
ax = fig.add_subplot(111, projection='3d')
ax.add_collection3d(Poly3DCollection(verts_cube, facecolors='sienna', edgecolors='black', linewidths=.8, alpha=0.2))
ax.add_collection3d(Poly3DCollection(verts_square, facecolors='b', edgecolors='b', linewidths=1.5, alpha=0.1, label="infiltration area"))
ax.scatter(el_x, el_y, el_z, marker='o', color='r', s=5, label="electrode")
ax.scatter(fdr_points[:, 0], fdr_points[:, 1], fdr_points[:, 2], color='k', marker='*', s=20, label="FDR sensor")
ax.add_collection3d(Poly3DCollection(verts_parallelepiped, facecolors='tab:brown', edgecolors='tab:brown', linewidths=.3, alpha=0.3, label="compaction"))
# Calcola la coordinata centrale di x nel parallelepipedo
x_center = (x_min + x_max) / 2
ax.plot([x_center, x_center], [y_min, y_max], [0, 0], color='magenta', linewidth=2, label="ERT line")


# Cosmetic settings
ax.set_xlim([-.8, 1.8])
ax.set_ylim([-.8, 1.8])
ax.set_zlim([-2, 0]) 
ax.set_xlabel("x (m)",fontsize=9)
ax.set_ylabel("y (m)",fontsize=9)
ax.set_zlabel("depth (m)",fontsize=9)
#plt.title('HC: high compaction',fontsize=10)
ax.legend(
    loc='upper right',        # Posizione di base della legenda
    fontsize=10,
    frameon=True,
    borderpad=0.2,
    labelspacing=0.3,
    bbox_to_anchor=(1.1, 1.1)  # Spostare la legenda in alto a destra, un po' sopra
)

plt.subplots_adjust(left=0.1, right=0.9, top=0.9, bottom=0.1)
fig.savefig(figdir+ 'scheme_HIGHcomp.png', dpi=200,)
            #transparent=True)
plt.show()
```
