```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Dominio
# -----------------------------
nx, nz = 80, 40
x = np.linspace(-10, 10, nx)
z = np.linspace(0, -5, nz)  # z negativa verso il basso
X, Z = np.meshgrid(x, z)

# -----------------------------
# Conducibilità idraulica K
# -----------------------------
K = np.ones_like(X)

# zona compattata centrale
compacted = (np.abs(X) < 3) & (Z > -2)
K[compacted] = 0.05   # K molto più bassa

# aumento di K ai lati con la profondità (più saturi)
K += 0.3 * np.exp(Z)

# -----------------------------
# Carico idraulico H (concettuale)
# -----------------------------
H = Z.copy()  # termine gravitazionale

# accumulo di carico sopra la zona compattata
H += 1.2 * np.exp(-(X / 2)**2) * np.exp(Z / 1.2)

# dissipazione laterale (dove K è alta)
H -= 0.5 * np.exp(-((X - 6)/3)**2) * np.exp(Z / 1.5)
H -= 0.5 * np.exp(-((X + 6)/3)**2) * np.exp(Z / 1.5)

# -----------------------------
# Plot 3D
# -----------------------------
fig = plt.figure(figsize=(14, 5))

# ---- K ----
ax1 = fig.add_subplot(121, projection='3d')
surf1 = ax1.plot_surface(X, Z, K, cmap='viridis')
ax1.set_title("Conducibilità idraulica K")
ax1.set_xlabel("x")
ax1.set_ylabel("z")
ax1.set_zlabel("K")
fig.colorbar(surf1, ax=ax1, shrink=0.5)

# ---- H ----
ax2 = fig.add_subplot(122, projection='3d')
surf2 = ax2.plot_surface(X, Z, H, cmap='plasma')
ax2.set_title("Carico idraulico H")
ax2.set_xlabel("x")
ax2.set_ylabel("z")
ax2.set_zlabel("H")
fig.colorbar(surf2, ax=ax2, shrink=0.5)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

# Dominio
# -----------------------------
nx, nz = 120, 60
x = np.linspace(-10, 10, nx)
z = np.linspace(0, -1.2, nz)   # profondità fino a 1.2 m
X, Z = np.meshgrid(x, z)

# Conducibilità idraulica satura Ks [cm/day]
# -----------------------------
# Suolo non compattato:
# 1000 cm/day in superficie -> 500 cm/day a 1 m
Ks = 1000 - 500 * (-Z / 1.0)
Ks[Ks < 500] = 500
# Zona compattata centrale (fino a 25 cm)
compacted = (np.abs(X) < 3) & (Z > -0.25)
Ks[compacted] = 100.0

# Carico idraulico H [m]
# -----------------------------
# Termine gravitazionale
H = Z.copy()
# Accumulo sopra lo strato compattato
H += 0.25 * np.exp(-(X / 2.0)**2) * np.exp(Z / 0.25)
# Dissipazione laterale
H -= 0.15 * np.exp(-((X - 6)/3)**2) * np.exp(Z / 0.5)
H -= 0.15 * np.exp(-((X + 6)/3)**2) * np.exp(Z / 0.5)



# Plot
plt.figure(figsize=(6,4))
# Ks come sfondo
im = plt.pcolormesh(X, Z, Ks, cmap='viridis', shading='auto')
cbar = plt.colorbar(im, shrink=0.7)
cbar.set_label(r"$K_s$ [cm/day]")
# Linee di carico idraulico
levels = np.linspace(H.min(), H.max(), 10)
cs = plt.contour(X, Z, H, levels=levels, colors='r', linewidths=1.2)
plt.clabel(cs, fmt="%.2f m", fontsize=8)

# Zoa compattata
plt.axvspan(-3, 3, ymin=0.79, ymax=1.0, color='darkviolet', alpha=0.1)

# Area di infiltrazione (linea blu in superficie)
plt.plot(x, np.zeros_like(x)+0.01, color='blue', lw=10)
plt.text(0, 0.03, "Area di infiltrazione", color='blue',
         ha='center', va='bottom', fontsize=10)

# Legenda
legend_elements = [
    Line2D([0], [0], color='r', lw=1.5, ls='--', label="Carico idraulico"),
    Line2D([0], [0], color='blue', lw=3, label="Area di infiltrazione"),
    Line2D([0], [0], color='darkviolet', lw=6, alpha=0.2, label="Zona compattata")
]
plt.legend(handles=legend_elements, loc="lower left")

plt.xlabel("x [m]")
plt.ylabel("z [m]")
plt.ylim(-1, None)
plt.title("Flusso preferenziale indotto da strato compattato\n")
plt.tight_layout()
plt.savefig('comp.png', dpi=150)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

# -----------------------------
# Dominio (RISCALE)
# -----------------------------
nx, nz = 120, 60
x = np.linspace(-2, 2, nx)
z = np.linspace(0, -1.1, nz)
X, Z = np.meshgrid(x, z)

# -----------------------------
# Ks NON compattato [cm/day]
# -----------------------------
Ks_base = 1000 - 500 * (-Z / 1.0)
Ks_base[Ks_base < 500] = 500

# -----------------------------
# Ks CON compattazione
# -----------------------------
Ks_comp = Ks_base.copy()
compacted = (np.abs(X) < 0.6) & (Z > -0.25)
Ks_comp[compacted] = 100.0

# -----------------------------
# Carico idraulico H [m]
# -----------------------------
# Caso NON compattato (campo omogeneo)
H_hom = Z.copy()

# Caso CON compattazione
H_comp = Z.copy()
H_comp += 0.25 * np.exp(-(X / 0.5)**2) * np.exp(Z / 0.25)
H_comp -= 0.15 * np.exp(-((X - 1.2)/0.8)**2) * np.exp(Z / 0.5)
H_comp -= 0.15 * np.exp(-((X + 1.2)/0.8)**2) * np.exp(Z / 0.5)

# Livelli comuni di carico
levels = np.linspace(min(H_hom.min(), H_comp.min()),
                     max(H_hom.max(), H_comp.max()), 10)

# -----------------------------
# Plot
# -----------------------------
fig, axes = plt.subplots(1, 2, figsize=(9, 4), sharey=True)

# ===== SINISTRA: NON COMPATTATO =====
ax = axes[0]
im0 = ax.pcolormesh(X, Z, Ks_base, cmap='viridis', shading='auto')
cs0 = ax.contour(X, Z, H_hom, levels=levels, colors='r', linewidths=1.2)
ax.clabel(cs0, fmt="%.2f m", fontsize=8)

ax.plot(x, np.zeros_like(x)+0.03, color='blue', lw=8, alpha=.4)
ax.set_title("Suolo omogeneo non compattato")
ax.set_xlabel("x [m]")
ax.set_ylabel("z [m]")
ax.set_ylim(-1.0, None)

cbar0 = plt.colorbar(im0, ax=ax, shrink=0.8)
cbar0.set_label(r"$K_s$ [cm/day]")

# ===== DESTRA: COMPATTATO =====
ax = axes[1]
im1 = ax.pcolormesh(X, Z, Ks_comp, cmap='viridis', shading='auto')
cs1 = ax.contour(X, Z, H_comp, levels=levels, colors='r', linewidths=1.2)
ax.clabel(cs1, fmt="%.2f m", fontsize=8)

ax.axvspan(-0.6, 0.6, ymin=.95, ymax=0.93, color='b')
ax.plot(x, np.zeros_like(x)+0.03, color='blue', lw=8, alpha=.4)

ax.set_title("Suolo con layer compattato")
ax.set_xlabel("x [m]")

cbar1 = plt.colorbar(im1, ax=ax, shrink=0.8)
cbar1.set_label(r"$K_s$ [cm/day]")

# -----------------------------
# Legenda (solo a destra)
# -----------------------------
legend_elements = [
    Line2D([0], [0], color='r', lw=1.5, label="Carico idraulico"),
    Line2D([0], [0], color='b', lw=6, alpha=0.4, label="Area di infiltrazione"),
    Line2D([0], [0], color='blue', lw=4, label="Acqua stagnante")
    
]
axes[1].legend(handles=legend_elements, loc="lower left")
plt.tight_layout()
plt.savefig("infiltr_comp.png", dpi=150)
plt.show()
```
